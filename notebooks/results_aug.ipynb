# Результаты эксперимента с аугментациями

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
metrics = pd.read_csv('../logs/metrics.csv')
```

```python
metrics.tail()
```

```python
plt.title('Test accuracy')
sns.lineplot(x='round', y='test_accuracy', data=metrics)
plt.show()
```

```python
plt.title('Test F1')
sns.lineplot(x='round', y='test_f1', data=metrics)
plt.show()
```
