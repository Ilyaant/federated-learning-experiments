# Результаты эксперимента без аугментаций

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
metrics = pd.read_csv('../logs/exp_no_aug/metrics.csv')
```

```python
metrics.tail()
```

## Результаты на test

```python
plt.title('Test accuracy')
sns.lineplot(x='round', y='test_accuracy', data=metrics)
plt.show()
```

```python
plt.title('Test F1')
sns.lineplot(x='round', y='test_f1', data=metrics)
plt.show()
```

## Результаты на train

```python
plt.title('Train accuracy')
sns.lineplot(x='round', y='train_accuracy', data=metrics)
plt.show()
```

```python
plt.title('Train F1')
sns.lineplot(x='round', y='train_f1', data=metrics)
plt.show()
```

## Совмещенные графики

```python
plt.title('Accuracy')
sns.lineplot(x='round', y='train_accuracy', data=metrics, label='train')
sns.lineplot(x='round', y='test_accuracy', data=metrics, label='test')
sns.lineplot(x='round', y='val_accuracy', data=metrics, label='val')
plt.ylabel('accuracy')
plt.show()
```

```python
plt.title('F1')
sns.lineplot(x='round', y='train_f1', data=metrics, label='train')
sns.lineplot(x='round', y='test_f1', data=metrics, label='test')
sns.lineplot(x='round', y='val_f1', data=metrics, label='val')
plt.ylabel('F1')
plt.show()
```
